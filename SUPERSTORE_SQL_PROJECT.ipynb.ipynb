```python
import pandas as pd
import sqlite3

df = pd.read_csv('train.csv', encoding='latin1')


df.columns = df.columns.str.replace(' ', '_').str.replace('-', '_')
print("Columns:", df.columns.tolist())
print("\nShape:", df.shape)

conn = sqlite3.connect('superstore.db')
df.to_sql('orders', conn, if_exists='replace', index=False)
print("\nDatabase ready! ✅\n")

print("--- SQL PROJECT RESULTS ---\n")

# Now column names are like Customer_Name, Product_Name etc
q1 = pd.read_sql("SELECT SUM(Sales) as Total_Sales FROM orders", conn)
print("1. Total Sales"); print(q1, "\n")

q2 = pd.read_sql("SELECT Category, SUM(Sales) as Sales FROM orders GROUP BY Category", conn)
print("2. Sales by Category"); print(q2, "\n")

q3 = pd.read_sql("SELECT Region, SUM(Sales) as Sales FROM orders GROUP BY Region ORDER BY Sales DESC", conn)
print("3. Sales by Region"); print(q3, "\n")

q4 = pd.read_sql("SELECT Customer_Name, SUM(Sales) as Sales FROM orders GROUP BY Customer_Name ORDER BY Sales DESC LIMIT 5", conn)
print("4. Top 5 Customers"); print(q4, "\n")

q5 = pd.read_sql("SELECT Sub_Category, SUM(Sales) as Sales FROM orders GROUP BY Sub_Category ORDER BY Sales DESC LIMIT 5", conn)
print("5. Top Sub-Category by Sales"); print(q5, "\n")

q6 = pd.read_sql("SELECT State, SUM(Sales) as Sales FROM orders GROUP BY State ORDER BY Sales DESC LIMIT 5", conn)
print("6. Top 5 States"); print(q6, "\n")

q7 = pd.read_sql("SELECT Ship_Mode, COUNT(*) as Orders FROM orders GROUP BY Ship_Mode", conn)
print("7. Orders by Ship Mode"); print(q7, "\n")

q8 = pd.read_sql("SELECT Product_Name, SUM(Sales) as Sales FROM orders GROUP BY Product_Name ORDER BY Sales DESC LIMIT 5", conn)
print("8. Top Selling Products"); print(q8)
```
